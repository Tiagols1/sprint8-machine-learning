```python
import pandas as pd

# Carregando os dados do arquivo correto
df = pd.read_csv('users_behavior.csv')

# Exibindo as primeiras linhas para confirmar que funcionou
display(df.head())

# Exibindo informações gerais e o tamanho do dataset
print("\nInformações do DataFrame:")
df.info()
```

```python
# Verificando a distribuição dos planos (is_ultra)
target_counts = df['is_ultra'].value_counts(normalize=True) * 100
print("Distribuição da coluna alvo (is_ultra) em percentual:")
print(f"Plano Smart (0): {target_counts.get(0, 0):.2f}%")
print(f"Plano Ultra (1): {target_counts.get(1, 0):.2f}%")
```

Análise Exploratória Inicial:

- O dataset possui 3.214 linhas e 5 colunas (calls, minutes, messages, mb_used, is_ultra).
- Uma verificação de valores ausentes confirmou que não há dados nulos no arquivo[cite: 2].
- A coluna alvo (`is_ultra`) mostra a distribuição dos planos: 69.35% dos usuários utilizam o plano Smart (0) e 30.65% utilizam o plano Ultra (1)[cite: 2]. 
- Isso nos fornece uma *baseline* útil de aproximadamente 69.35% de acurácia caso o modelo previsse sempre a classe majoritária.

<div class="alert alert-block alert-warning">
Seu resumo agora está sustentado por resultados reais: a divisão de 69,35% / 30,65% vem do seu próprio cálculo, e o baseline da classe majoritária de cerca de 69,35% é uma boa forma de contextualizar a acurácia dos seus modelos. Dois tópicos ainda contêm artefatos `[cite: 2]` que parecem sobras de copiar e colar. Apague-os para que o texto fique limpo.
</div>

```python
# Separando features e target
features = df.drop(['is_ultra'], axis=1)
target = df['is_ultra']

# 1. Separando 60% para treino e 40% temporário (com stratify para manter a proporção das classes)
features_train, features_temp, target_train, target_temp = train_test_split(
    features, target, test_size=0.4, random_state=54321, stratify=target
)

# 2. Dividindo os 40% temporários igualmente em validação e teste (20% cada)
features_valid, features_test, target_valid, target_test = train_test_split(
    features_temp, target_temp, test_size=0.5, random_state=54321, stratify=target_temp
)

print("Tamanho do conjunto de treino:", len(features_train))
print("Tamanho do conjunto de validação:", len(features_valid))
print("Tamanho do conjunto de teste:", len(features_test))
```

<div class="alert alert-block alert-success">
Ótimo trabalho ao usar `stratify` nas duas etapas de divisão. Isso mantém as proporções de Smart/Ultra semelhantes nos três conjuntos, o que é importante quando o alvo é desbalanceado. A divisão 60/20/20 (1928 / 643 / 643) é sensata para cerca de 3.200 linhas, e o `random_state` fixo a torna reproduzível.
</div>

```python
best_tree_model = None
best_tree_acc = 0
best_depth = 0

# Aumentamos o range para testar profundidades até 15
for depth in range(1, 16):
    model = DecisionTreeClassifier(random_state=54321, max_depth=depth)
    model.fit(features_train, target_train)
    predictions = model.predict(features_valid)
    acc = accuracy_score(target_valid, predictions)
    
    if acc > best_tree_acc:
        best_tree_acc = acc
        best_tree_model = model
        best_depth = depth

print(f"Melhor Acurácia (Árvore de Decisão): {best_tree_acc:.4f} com max_depth = {best_depth}")
```

<div class="alert alert-block alert-success">
Bom trabalho ao ampliar o intervalo de `max_depth` para a árvore de decisão. A melhor profundidade (7, acurácia de validação 0.7916) agora está dentro do intervalo, em vez de estar na borda, então a busca é mais convincente. A árvore é selecionada usando apenas a acurácia de validação, que é a abordagem correta.
</div>

```python

best_rf_model = None
best_rf_acc = 0
best_est = 0
best_rf_depth = 0

# Ampliando o leque de n_estimators (até 60) e max_depth (até 15)
for est in range(10, 61, 10):
    for depth in range(1, 16):
        model = RandomForestClassifier(random_state=54321, n_estimators=est, max_depth=depth)
        model.fit(features_train, target_train)
        predictions = model.predict(features_valid)
        acc = accuracy_score(target_valid, predictions)
        
        if acc > best_rf_acc:
            best_rf_acc = acc
            best_rf_model = model
            best_est = est
            best_rf_depth = depth

print(f"Melhor Acurácia (Floresta Aleatória): {best_rf_acc:.4f} com n_estimators = {best_est} e max_depth = {best_rf_depth}")
```

<div class="alert alert-block alert-success">
Bom trabalho ao estender o grid sobre `n_estimators` e `max_depth` para a random forest. A melhor combinação (`n_estimators = 30`, `max_depth = 14`, acurácia de validação 0,8134) está dentro dos intervalos da sua busca e supera claramente a árvore única. Manter o objeto do melhor modelo para a avaliação final é uma boa prática.
</div>
